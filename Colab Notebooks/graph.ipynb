```python
import networkx as nx
import matplotlib.pyplot as plt

G_sym=nx.Graph()
G_sym.add_edge('Steven','Laura')
G_sym.add_edge('Steven','Marc')
G_sym.add_edge('Steven','John')
G_sym.add_edge('Steven','Michael')
G_sym.add_edge('Michael','Laura')
G_sym.add_edge('Marc','Laura')
G_sym.add_edge('Michael','John')
G_sym.add_edge('Steven','George')
G_sym.add_edge('Marc','John')
G_sym.add_edge('George','John')
G_sym.add_edge('George','Marc')
G_sym.add_edge('Marc','Marc')
plt.figure(figsize=(4,4))
nx.draw_networkx(G_sym)
```

```python
G_asym=nx.DiGraph()
G_asym.add_edge('A','B')
G_asym.add_edge('A','C')
G_asym.add_edge('A','D')
G_asym.add_edge('D','B')
G_asym.add_edge('C','D')
G_asym.add_edge('E','D')
G_asym.add_edge('D','A')
G_asym.add_edge('C','C')
plt.figure(figsize=(4,4))
nx.draw_networkx(G_asym)
```

```python
nx.spring_layout(G_asym)
plt.figure(figsize=(4,4))
nx.draw_networkx(G_asym)
```
