```python
import networkx as nx
import matplotlib.pyplot as plt
G_t=nx.read_edgelist('twitter_combined.txt.gz',create_using=nx.Graph(),nodetype = int)
plt.figure(figsize=(7,7))
nx.draw_networkx(G_t)
```

```python
pos = nx.spring_layout(df)
betCent = nx.betweenness_centrality(df, normalized=True,
endpoints=True)
node_color = [20000.0 * df.degree(v) for v in df]
node_size = [v * 10000 for v in betCent.values()]
plt.figure(figsize=(20,20))
nx.draw_networkx(df, pos=pos, with_labels=False,node_color=node_color,node_size=node_size )
plt.axis('off');
```
